```python
email_df["subject"] = email_df["subject"].fillna("")
email_df["body"] = email_df["body"].fillna("")
```

```python
email_df["text"] = (
    email_df["subject"].astype(str) + " " +
    email_df["body"].astype(str)
)
```

```python
email_df = email_df[email_df["text"].str.strip() != ""].copy()
```

```python
email_df = email_df.drop_duplicates(subset=["text"]).copy()
```

```python
print("Dataset shape:", email_df.shape)
print("\nClass distribution:")
print(email_df["label"].value_counts())
```

```python
X = email_df["text"]
y = email_df["label"]
```

```python
print("Number of emails:", len(X))
print("Number of labels:", len(y))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    max_features=10000,
    ngram_range=(1, 2),
    min_df=2
)
```

```python
X_train_tfidf = tfidf.fit_transform(X_train)
```

```python
X_test_tfidf = tfidf.transform(X_test)
```

```python
print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)
```

```python
from sklearn.svm import LinearSVC
```

```python
svm_model = LinearSVC(random_state=42)

svm_model.fit(X_train_tfidf, y_train)
```

```python
#make predictions
svm_pred = svm_model.predict(X_test_tfidf)
```

```python
#train the model
svm_model = LinearSVC(random_state=42)

svm_model.fit(X_train_tfidf, y_train)
```

```python
#make predictions
svm_pred = svm_model.predict(X_test_tfidf)
```

```python
#random forest
from sklearn.ensemble import RandomForestClassifier
```

```python
#create and train
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf_model.fit(X_train_tfidf, y_train)
```

```python
#predict
rf_pred = rf_model.predict(X_test_tfidf)
```

```python
#XGBoost(import)
from xgboost import XGBClassifier
```

```python
#Create the model;
xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)
```

```python
#train
xgb_model.fit(X_train_tfidf, y_train)
```

```python
#predict
xgb_pred = xgb_model.predict(X_test_tfidf)
```

```python
#evaluate the evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
#create a function
def evaluate_model(name, y_true, y_pred):
    print(f"\n{'=' * 50}")
    print(name)
    print(f"{'=' * 50}")
    
    print(f"Accuracy : {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision: {precision_score(y_true, y_pred):.4f}")
    print(f"Recall   : {recall_score(y_true, y_pred):.4f}")
    print(f"F1-score : {f1_score(y_true, y_pred):.4f}")
    
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))
    
    print("Confusion Matrix:")
    print(confusion_matrix(y_true, y_pred))
```

```python
#evaluate
evaluate_model("Support Vector Machine", y_test, svm_pred)
```

```python
evaluate_model("Random Forest", y_test, rf_pred)
```

```python
evaluate_model("XGBoost", y_test, xgb_pred)
```

```python
#compare the three models
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

results = pd.DataFrame({
    "Model": [
        "Support Vector Machine",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        accuracy_score(y_test, svm_pred),
        accuracy_score(y_test, rf_pred),
        accuracy_score(y_test, xgb_pred)
    ],
    "Precision": [
        precision_score(y_test, svm_pred),
        precision_score(y_test, rf_pred),
        precision_score(y_test, xgb_pred)
    ],
    "Recall": [
        recall_score(y_test, svm_pred),
        recall_score(y_test, rf_pred),
        recall_score(y_test, xgb_pred)
    ],
    "F1-score": [
        f1_score(y_test, svm_pred),
        f1_score(y_test, rf_pred),
        f1_score(y_test, xgb_pred)
    ]
})

results = results.round(4)

results
```

```python
#Calculate false negative
def get_confusion_metrics(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    
    return {
        "True Negatives": tn,
        "False Positives": fp,
        "False Negatives": fn,
        "True Positives": tp
    }


svm_cm = get_confusion_metrics(y_test, svm_pred)
rf_cm = get_confusion_metrics(y_test, rf_pred)
xgb_cm = get_confusion_metrics(y_test, xgb_pred)

comparison = pd.DataFrame(
    [svm_cm, rf_cm, xgb_cm],
    index=["SVM", "Random Forest", "XGBoost"]
)

comparison
```

```python
#save comparism results
results.to_csv("../reports/model_comparison.csv", index=False)

comparison.to_csv("../reports/confusion_matrix_metrics.csv")

print("Model comparison saved.")
```

```python
import pandas as pd
import joblib
from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ============================================================
# 1. LOAD EXISTING DATASET
# ============================================================

df = pd.read_csv("../data/raw/CEAS_08.csv")

print("=" * 70)
print("DATASET")
print("=" * 70)
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("\nLabel distribution:")
print(df["label"].value_counts().sort_index())


# ============================================================
# 2. LOAD EXISTING SAVED MODELS
# ============================================================

tfidf = joblib.load("../models/tfidf_vectorizer.pkl")
svm_model = joblib.load("../models/svm_model.pkl")
rf_model = joblib.load("../models/rf_model.pkl")
xgb_model = joblib.load("../models/xgb_model.pkl")

print("\n" + "=" * 70)
print("SAVED MODELS")
print("=" * 70)
print("TF-IDF vectorizer: loaded")
print("SVM model: loaded")
print("Random Forest model: loaded")
print("XGBoost model: loaded")


# ============================================================
# 3. IDENTIFY THE TEXT COLUMN
# ============================================================

df["text"] = (
    df["subject"].fillna("").astype(str)
    + " "
    + df["body"].fillna("").astype(str)
)

X = df["text"]
y = df["label"]

X_tfidf = tfidf.transform(X)


# ============================================================
# 4. EVALUATE THE EXISTING MODELS
# ============================================================

models = {
    "SVM": svm_model,
    "Random Forest": rf_model,
    "XGBoost": xgb_model
}

print("\n" + "=" * 70)
print("MODEL EVALUATION")
print("=" * 70)

for name, model in models.items():

    predictions = model.predict(X_tfidf)

    accuracy = accuracy_score(y, predictions)
    precision = precision_score(y, predictions, zero_division=0)
    recall = recall_score(y, predictions, zero_division=0)
    f1 = f1_score(y, predictions, zero_division=0)

    print(f"\n{name}")
    print("-" * 40)
    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1-score : {f1:.4f}")

    print("\nConfusion Matrix:")
    print(confusion_matrix(y, predictions))

    print("\nClassification Report:")
    print(classification_report(
        y,
        predictions,
        target_names=["Legitimate", "Phishing"],
        zero_division=0
    ))


# ============================================================
# 5. VERIFY THE LABEL MAPPING USED BY THE DATASET
# ============================================================

print("\n" + "=" * 70)
print("LABEL MAPPING USED IN THE PROJECT")
print("=" * 70)
print("0 = Legitimate")
print("1 = Phishing/Spam")


# ============================================================
# 6. SAMPLE PREDICTION TEST
# ============================================================

test_email = """
Congratulations! You have won a special prize.
Click the link below to claim your reward immediately.
"""

test_vector = tfidf.transform([test_email])

print("\n" + "=" * 70)
print("SAMPLE PREDICTION")
print("=" * 70)

for name, model in models.items():

    prediction = model.predict(test_vector)[0]

    if prediction == 1:
        result = "Phishing"
    else:
        result = "Legitimate"

    print(f"{name}: {result}")

print("\nML verification completed.")
```
